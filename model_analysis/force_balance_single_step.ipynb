# Mathematical background & analysis

In this notebook we explore how horizontal force balance is realised in a numerical subduction model, with a particular focus on how slab buoyancy is transmitted to the trailing plate.

The model analysed here is a 2-D Stokes-flow (creeping-flow) model, so it already satisfies force and moment balance to a very good approximation (see ./integration_residuals.ipynb). The aim of this notebook is therefore not to solve the force balance, but to reconstruct and interpret it over finite domains by integrating traction components.

We begin by validating the **basic form** of the vertically integrated horizontal force balance. We then decompose the horizontal normal stress into components associated with a) topography and density structure, b) deviatoric stress (expressed in the in-plane differential resulatant).

To develop our representation of the vertically-integrated force balance, we will consider a rectangular sub-domain whose boundaries are aligned with the coordinate axes.

----


![Schematic of force balance](figures/domain.png)

**Figure 1.** Schematic illustration of the horizontal force balance.

----

In this geometry, the traction vector on each boundary has only a single non-zero horizontal component. On the vertical boundaries ($\Omega_{0,1}$), the outward normal is horizontal, so the relevant traction component is:

$t_x = \sigma_{xx}$. 


On the basal boundary ($\Omega_{2}$), where the outward normal is vertical, the horizontal traction is $t_x = \tau_{zx} $.


**vertically integrated force balance - basic form**


Integrating the static equilibrium condition over the three contributing boundaries therefore reduces the vector force balance to a scalar equation for the horizontal force per unit length:


\begin{equation}
\sum F_x = \bar \sigma_{xx}(x_1) - \bar \sigma_{xx}(x_0) + \int_{x_0}^{x_1} \tau_{zx}(x, z_c) dx = 0 \;\;\mathrm{{\bf(1a)}}
\end{equation}

The **overbars** in Equation **(1)** denote vertical integration between an upper equipotential surface—e.g. sea level—and a lower equipotential surface which is often referred to as the depth of compensation ($z_c$). This depth is somewhat arbitrary, but the assumption is that the depth is sufficiently large that horizontal gradients in pressure (or vertical normal stress) are negligible$\bf{^1}$. Under the hydrostatic approximation, there are no pressure gradients along equipotentials -  this is a typical assumption used in thin-plate flexural models of the downbending subducting plate.


Using the $\Delta$ symbol to represent the difference between the vertical integrals at points $x_0$ and $x_1$, we can simplify further: 


\begin{equation}
\Delta \bar \sigma_{xx}+ \int_{x_0}^{x_1} \tau_{zx}(x, z_c) dx = 0 \;\;\mathrm{{\bf(1b)}}
\end{equation}


Equation **(1)** can be interpreted as follows: the difference in integrated horizontal traction due to basal shear stress ($\tau_{zx}$) between $x_0$ and $x_1$, at $z = z_c$, balances the difference in the vertically integrated horizontal traction due to the normal stress $\sigma_{xx}$. 

$\bf{^1}$ Equation **(1)** remains valid even if pressure gradients exist along the basal boundary. However, a significant pressure gradient implies "dynamic" support of the topography and complicates the interpretation of the terms. The analysis here shows that dynamic topography in the subduction model is negligible - at least on the horizontal scale of the trench topography. 

**incremental form vertically integrated force balance**

The equation above is derived for two arbitrary points ($x_0$, $x_1$), and simply states that these quantities must cancel — a condition of static equilibrium (horizontal force balance). The points are of course arbitrary, but our interest here is the effect of the trench (Earth's topographic minima) relative to an isostatic column in the trailing plate.

If we fix $x_0$ (e.g. at the trench) and allow $x_1$ to be a variable, then we can regard these quantities in Equation 2 as **functions of** $x$.  In the plots that follow, we now define the $\Delta$ operator to represent $\Delta(f,x_0,x) = f(x) - f(x_0)$. 

...so that differences are always measured relative to the reference position x_0. 

This **incremental form** is what we will acutally plot. 


**vertically integrated force balance - interesting form**


As often done in thin-sheet models of the lithosphere (and ice-shelves etc), we can rewrite $\sigma_{xx}$ as: 

$\sigma_{xx} = \tau_{xx} + \sigma_I$

where $\sigma_I = -P$ is the mean stress and stresses are taken as negative in compression.

In a vertically stratified system, the vertical normal stress $\sigma_{zz}$ plays a central role in determining the mean stress, because it reflects the integrated weight of the overlying material. Specifically, $\sigma_{zz}$ is controlled by both the density structure of the column and its surface and interface topography. Writing the deviatoric vertical stress as:


$\tau_{zz} = \sigma_{zz} - \sigma_I$

we can equivalently express the horizontal normal stress as

$\sigma_{xx} = \tau_{xx} + \sigma_{zz} - \tau_{zz}$

Substituting in Equation **(1)** gives:

\begin{equation}
\Delta (\overline{\tau_{xx}-\tau_{zz}})  +  \Delta \bar\sigma_{zz}  +  \int_{x_0}^{x_1} \tau_{zx}(x, z_c) = 0 \;\;\mathrm{{\bf(2)}}
\end{equation}

------

**Why choose this representation?**

It is important to recognise that the vertical stress components are being used here as proxies for the pressure field. Vertical stresses only influence the horizontal force balance via pressure — it’s the pressure field that couples vertical loading to horizontal forces (drop a bowl of soup on the floor if you want a very direct demonstration of how vertical stresses generate horizontal pressure gradients & motion).

Taken together, the vertical stress components provide a convenient representation of pressure that allows us to (broadly speaking) separate the effects of topography and density structure from pressure perturbations associated with in-plane differential stress. In this decomposition, the former are carried primarily by $\sigma_{zz}$,

The latter are expressed through the deviatoric stresses. Indeed we can also write:

$\tau_{xx} - \tau_{zz} \;\equiv\; \tau_{xx} + P’$

where $P’$ is commonly referred to as the tectonic overpressure. This quantity represents the departure of the pressure field from its lithostatic reference state arising from tectonic stress, rather than from variations in topography or density structure.



**Interpreting force terms and directions**


In Equation **(1)** & **(2)** a positive change in any of the quantities, from $x_0 \rightarrow x_1$ represents a net force to the right. 

For instance a positive change in $\Delta (\overline{\tau_{xx}-\tau_{zz}})$ represents a reduction in the integrated horizontal normal stress relative to the vertical, we can say that the in-plane differential stress resultant is becoming less compressional in the positive $x$ direction.  

In the manuscript, these 3 terms are symbolised: 

\begin{equation}
\Delta N_D  -  \Delta \mathrm{GPE}^*  +  F_B = 0 \;\; \mathrm{{\bf(3)}}
\end{equation}

Where the definition imposed is: $\Delta \mathrm{GPE}^* = -\Delta \bar\sigma_{zz}$

The sign change implemented here makes the sign of $\Delta \mathrm{GPE}^* $ consistent with the definition for the "true" GPE (which is based on positive quantity - pressure, rather than a negative quantity, the vertical normal stress).  This means that a positive change in $\Delta \mathrm{GPE}^*$ from  $x_0 \rightarrow x_1$, now represents a net force to the left. This actually turns out to be quite useful in plotting variation in terms - as we will see later. 

The reason for the asterisk on the $\mathrm{GPE}^*$, is discussed in the paper. Wherever we refer to $\mathrm{GPE}^*$ (asterisk included) we are referring to the (negative of) the vertical integral of "true" vertical normal stress $\bar\sigma_{zz}$. Meanwhile, the "true" $\mathrm{GPE}$ is the integral of the lithostatic pressure ($\bar P_L$). In general, $\mathrm{GPE}^* \neq \mathrm{GPE}$. The quantities are only equal when the vertical normal stress in a column is lithostatic. This is not the case for trench topography!. The paper shows that one can interpret  $\bar\sigma_{zz}$ in as a GPE-like quantity, even for non-isostatic columns, if we use a corrected density ($\rho^*$) that accounts for horizontal gradients in the vertical shear stress: $\frac{\partial \tau_{zx}}{\partial x}$.

**Some further points:**

- It is only the implied location of the columns in Equations **(1)** - **(3)** that differentiates the trench pull force, from e.g, ridge push; both are net horizontal force components that arise due to differences described by the quantity $\Delta \bar\sigma_{zz}$, produced by topographic changes.

- In an idealised 2D subducting plate, these 2 contributions (ridge push and trench pull) combine in a non-overlapping way: the ridge push captures the net force due to the isostatic topography, while the trench pull captures the net force  due to the non-isostatic topography. Each of the force components has a shared reference column, being an isostatic column of "old" lithosphere close to the trench. 

- The absolute value of $N_D$ has meaning: it defines (in an overall sense) whether the stress state is tending to extend $N_D>0$ or shorten $N_D<0$ the lithosphere. Because significant differential stress can be supported only where $z' < z'_m$, the value of $N_D$ (i.e., the integral) should converge to a constant value (approximately independent of $z_c$).

- The magnitude of $\mathrm{GPE}^*$ in a column has no direct significance in terms of the horizontal force balance - only differences between columns matter. Unlike $N_D$, the magnitude of $\mathrm{GPE}^* = \bar\sigma_{zz}$ in a single column always increases if we choose a larger integration depth. However, the integral of the $\Delta \mathrm{GPE}^* = \Delta \bar\sigma_{zz}$ (difference between columns) should converge with integration depth - if the assumption about the negligible pressure (or vertical normal stress) gradients at the compensation level is valid.

- In the MDOODZ7.0 code, the convention is that $z$ is positive up. In this notebook, the $z$ coordinate is switched for consistency with the manuscript. In terms of the signs of the stress tensor components, the only other required change in sign is for the shear stress.

- The origin of the z coordinate in this analysis is the initial level of the rock surface interface, the upper limit of the vertical integrations is the top of the model domain ($z=$ -20 km), inclusive of the "air" layer. The region above the rock surface has zero density, so $(\rho_m - \rho_w) = \rho_m =$ 3250 $\text{kg m}^{-3}$.


**Proposed scaling relationship:**


In the manuscript, the following scaling relationship is proposed for the trench pull force:  

\begin{equation}
\text{trench pull force} \equiv \Delta \mathrm{GPE}^* = (\rho_m - \rho_w) g w_T ( \frac{z'_m}{2}) \approx \Delta P_T z'_{np}
\end{equation}

Where $w_T$ is the deflection at the trench axis relative to the isostatic level, $\Delta P_T$ is referred to as the trench pressure deficit, $z'_m$ is the mechanical thickness, and $z'_{np}$ is the neutral plane depth. 

This notebook evaluates this relationship based on the results of a numerical model. Because the numerical model provides access to the vertical normal stress $\sigma_{zz}$, we can calculate the trench pull force  directly.

# Part 3: Horizontal force balance

We now use the stress tensor components from the numerical model to estimate the terms that appear in the **vertically integrated horizontal force balance** (equations 1 - 3).


First let's validate the basic form of the vertically-integrated force balance (i.e. Equation 1): 

\begin{equation}
\Delta \bar\sigma_{xx}(x) +  F_B(x) = 0 
\end{equation}


---

**Note: Discrete formulation on the finite-difference grid**

In the discrete system, each column corresponds to a vertical stack of **cells** (indexed by $(k = 0, 1, \dots, N_{x-1}$)), bounded by **vertices** ($N_x$ in total).   Vertical integrations are performed on quantities defined at **faces**, so integrated (barred) quantities (like `bar_sig_xx`) are naturally stored at **vertices**.

To form the discrete difference, we define $\Delta(\bar{f})$ on the **cell centers** $ (N_{x-1})$ elements, such that the value of  
$\Delta(\bar{f})$ at cell $k$ corresponds to:

$$
\Delta(\bar{f})_k = \bar{f}_{k+1} - \bar{f}_{ref},
\quad \text{for } k = 0, 1, \dots, N_x-2
$$


In other words, each cell-centered value $\Delta(\bar{f})_k$  represents the difference in vertically integrated traction between the traction on vertex index $k+1$  and a reference location $k_{ref}$. We are free to choose $k_{ref}$ wherever we like. Functionally, this choce simply ammounts to subtracting a constant from the overall force balacne - which remains valid. For analysing the variation of force components near the trench, it is practical to use the trench location as the refernce point.

---


```python
#-----------------------------------
#calculate the vertically integrated quantities
#-----------------------------------


# σ̄ₓₓ  — vertically integrated normal stress (Nz−1, Nx)
sig_xx_face = centers_to_vertical_faces(sig_xx_grid)
bar_sig_xx = np.cumsum(sig_xx_face * dz, axis=0)


# σ̄zz  ≡  −\mathrm{GPE}^*  — vertically integrated vertical normal stress
sig_zz_face = centers_to_vertical_faces(sig_zz_grid)
bar_sig_zz = np.cumsum(sig_zz_face * dz, axis=0)


# (τ̄ₓₓ − τ̄zz)  ≡  Fᴰ(x)  — vertically integrated differential stress

diff_stress_face = centers_to_vertical_faces(diff_stress_grid)
bar_diff_stress = np.cumsum(diff_stress_face * dz, axis=0)


#-----------------------------------
#calculate the basal shear term
#-----------------------------------

# ∫ τₓz dx ≡ Fᴮ(x)   — horizontally integrated basal shear traction
tau_zx_hface = vertices_to_horizontal_faces(tau_zx_grid)
FB_x_grid = np.cumsum(tau_zx_hface  * dx, axis=1)


#-----------------------------------
# Calcuate the Delta terms, by subtracting values at the reference location 
# (generically, x_0, here the trench)
#-----------------------------------


ref_index = tindx
del_bar_sig_xx      = bar_sig_xx[:, :-1]      - bar_sig_xx[:, [ref_index]]
del_bar_sig_zz      = bar_sig_zz[:, :-1]      - bar_sig_zz[:, [ref_index]]
del_bar_diff_stress = bar_diff_stress[:, :-1] - bar_diff_stress[:, [ref_index]]
FB_x_grid           = FB_x_grid[:]           - FB_x_grid[:, [ref_index]]



gpelabel = r'$\Delta \mathrm{GPE}^*(x)$' 
FD_label = r'$\Delta N_D(x)$'

```

```python
#this block construct the vertical shear stress resulatant (V) 
# and uses the condition dV/dx= 0 to establish the location of the first zero crossing (xi_loc)
# See additional notebook for plot so V, dV/dx and the flexural topography

tau_zx_vface = vertices_to_vertical_faces(tau_zx_grid)  # (Nz-1, Nx)
Vx = -1 * trapz(tau_zx_vface, dx=dz, axis=0)  # integrate over z, result shape (Nx,)
Vx = 0.5 * (Vx [:-1] + Vx [1:])
dVdx = np.gradient(Vx, dx)
vtopo = dVdx / (delrho * g)
window = 500  # search range in km (or adjusted if vxpts in m)
mask_x = vxpts > x_trench
tau_sum = np.sum(tau_zx_vface, axis=0)[mask_x]
indx = np.argmin(-1 * tau_sum[:window])
xi_loc = vxpts[mask_x][:window][indx]
xi_indx = np.argmin(np.abs(vxpts - xi_loc))
xi_dist = xi_loc - x_trench
xm_indx = int(tindx + 0.5 * (xi_indx - tindx))

#this is a depth mask to limit the search for the neutral plane, 
#or the zero-crossing of the function tau_xx - tau_zz 
np_mask = np.logical_and(czpts>10e3, czpts<50e3)

# Extract the masked array and corresponding x-values
z_vals = diff_stress_grid[:, xm_indx][np_mask]
x_vals = czpts[np_mask]

# Find indices where the sign changes (zero crossing)
sign_changes = np.where(np.diff(np.sign(z_vals)))[0]

# If a zero crossing exists, interpolate
if len(sign_changes) > 0:
    idx = sign_changes[0]  # Take the first crossing
    x1, x2 = x_vals[idx], x_vals[idx + 1]
    y1, y2 = z_vals[idx], z_vals[idx + 1]

    # Linear interpolation for zero crossing
    np_depth = x1 - y1 * (x2 - x1) / (y2 - y1)
    print(f"Interpolated neutral plane crossing at z = {1e-3*np_depth:.1f} km")
else:
    print("No zero crossing found within the specified range.")
```

## Plot the _interesting_ form of the horizontal force balance

We now want to look at the more _interesting_ form of the horizontal force balance, where the $\sigma_{xx}$ is decomposed in such a way as to highlight:

1. the pressure-component of the horizontal traction that is attributable to the depth distribution of the vertical normal stress ($\sigma_{zz}(z)$) (this is where topographic forces like swell push, ridge push and trench pull are expressed
2. the deviatoric-related component of the horizontal traction due to a finite differential stress resultant $(\tau_{xx} - \tau_{zz})$

In the symbolic form we write this as:

\begin{equation}
\Delta N_D  -  \Delta \mathrm{GPE}^*  +  F_B = 0 \;\; \mathrm{{\bf(3a)}}
\end{equation}

Note that for the overall force balance on a section of lithosphere bounded by any 2 columns, only the changes in $N_D$ and $\mathrm{GPE}^*$ are relevant - not the absolute value. For any valid force balance, we can add an arbitary consnant to the $N_D$ and $\mathrm{GPE}^*$, and the equilibroim is maintained.

## Final thoughts: applying this approach to other models

This notebook shows how integrating tractions over a finite sub-domain, together with a physically motivated decomposition of the horizontal normal stress, provides a clear way to diagnose horizontal force balance and to isolate the role of topography-controlled pressure anomalies.

If you want to apply this approach to other numerical subduction models, a few general points are worth noting.

* Topographic forces do not require explicit topography. Even in models with a free-slip upper surface, trench-pull– and ridge-push–type forces still appear, expressed as pressure perturbations along the top boundary. With a true free surface, stresses vanish at the surface; with free slip, normal velocities vanish but stresses remain. In both cases, ridges correspond to high pressure and trenches to low pressure. The key requirement is to integrate the actual model pressure, not a hypothetical lithostatic pressure derived from density alone.

* More generally, the equilibrium in a Stokes-flow model is fully described by the returned stress fields — pressure and deviatoric stress. Any force-balance analysis should therefore be based directly on these quantities. Integrating a notional lithostatic pressure can be a useful check, but it should not replace analysis of the true stresses.

* A simple and informative first step is to evaluate the differential stress resultant $N_D$ in a column beneath the trench. Integrating the deviatoric stresses and asking whether the result is positive, negative, or near zero already goes a long way. A large positive value (with compression taken as negative) would indicate tension-like coupling, whereas a neutral or compressional value rules it out. Comparing magnitudes to typical ridge-push estimates (∼2–4 $\mathrm{TN m}^{-1}$) provides useful context.

* Next, examine how this integrated quantity varies away from the trench. Variations of a few $\mathrm{TN m}^{-1}$, especially if they scale with trench-outer slope topography, are consistent with a pressure-controlled (topographic) force.

* Basal shear stresses can usually be assessed quickly. If basal shear at the compensation depth is of order a megapascal or less, it is unlikely to contribute meaningfully to the horizontal force balance over ∼100 km length scales.

Finally, once these elements are understood, constructing the full force balance as done here provides a consistent way to close the problem.
