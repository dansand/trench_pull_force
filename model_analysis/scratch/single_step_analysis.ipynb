# Mathematical background & analysis notes

In this analysis, we consider the horizontal force balance across a finite domain representing part of the trailing plate.

The domain is chosen such that its boundaries are parallel to the coordinate axes. In this configuration, the traction vector on each boundary has only one nonzero component aligned with the corresponding outward normal. For the vertical boundaries, the outward normal is horizontal, so the relevant traction component is $t_x = \sigma_{xx}$. For the basal boundary, where the normal is vertical, the traction component contributing to the horizontal force balance is $t_x = \tau_{xz}$.

As a result, the vector equation for static equilibrium reduces, upon integration over the 3 contrubuting domain boundaries, to a scalar equation for the x-component of the net force ($F_x$) per unit length:


\begin{equation}
\sum F_x = \bar \sigma_{xx}(x_1) - \bar \sigma_{xx}(x_0) + \int_{x_0}^{x_1} \tau_{xz}(x, z_c) dx = 0 \;\;\mathrm{{\bf(1a)}}
\end{equation}

The **overbars** in Equation **(1)** denote vertical integration between an upper equipotential surface—typically sea level, and a lower equipotential surface at the compensation depth ($z_c$). This depth is somewhat arbitrary, but it is commonly chosen such that horizontal gradients in pressure (or vertical normal stress) are negligible, i.e., under a hydrostatic approximation  $\bf{^1}$ . This is the standard assumption in thin-plate flexural models.


If we in $\Delta$ symbol is a shorthand used to represent the difference between the vertical integrals at points $x_0$ and $x_1$, we can simplify further: 


\begin{equation}
\Delta \bar \sigma_{xx}+ \int_{x_0}^{x_1} \tau_{xz}(x, z_c) dx = 0 \;\;\mathrm{{\bf(1b)}}
\end{equation}


Equation **(1)** can be interpreted as follows: the difference in integrated horizontal traction due to basal shear stress ($\tau_{xz}$) between $x_0$ and $x_1$, at $z = z_c$, balances the difference in the vertically integrated horizontal traction due to the normal stress $\sigma_{xx}$. 

 $\bf{^1}$ quation **(1)** remains valid even if pressure gradients exist along the basal boundary. However, a significant pressure gradient would implies dynamic support of the topography and complicate the interpretation of the terms. The analysis shows that dynamic topography in the subduction model is negligible - on teh horzontal sacke ofthe trench topotgrapht. This is consistent with the standard assumtpuin that trench topography anomaly is instead supported by vertical shear stresses within the lithosphere.


----------------

We can rewrite $\sigma_{xx}$ as: 

$\sigma_{xx} = \tau_{xx} + \sigma_I= \tau_{xx} + \sigma_{zz} - \tau_{zz}$

Where $\sigma_I = -P$ is the mean stress. Stress tensor quantities are assumed negative in compression. Substituting in Equation **(1)** gives:

\begin{equation}
\Delta (\overline{\tau_{xx}-\tau_{zz}})  +  \Delta \bar\sigma_{zz}  +  \int_{x_0}^{x_1} \tau_{xz}(x, z_c) = 0 \;\;\mathrm{{\bf(2)}}
\end{equation}

In Equation **(1)** & **(2)** a positive change in any of the quantities, from $x_0 \rightarrow x_1$ represents a net force to the right. For instance a positive change in $\Delta (\overline{\tau_{xx}-\tau_{zz}})$ represents a reduction in the integrated horizontal normal stress relative to the vertical, we can say that the in-plane stress is becoming less compressional in the positive $x$ direction.  

In the manuscript, these 3 terms are symbolised: 

\begin{equation}
\Delta F_{D}  -  \Delta \mathrm{GPE}^*  +  F_B = 0 \;\; \mathrm{{\bf(3)}}
\end{equation}

Where the definition imposed is: $\Delta \mathrm{GPE}^* = -\Delta \bar\sigma_{zz}$

The sign change implemented here makes the sign of $\Delta \mathrm{GPE}^* $ consistent with the definition for the "true" GPE (which is based on positive quantity - pressure, rather than a negative quantity, the vertical normal stress).  This means that a positive change in $\Delta \mathrm{GPE}^*$ from  $x_0 \rightarrow x_1$, now represents a net force to the left. The reason for the asterisk on the $\mathrm{GPE}^*$, is discussed in the paper. Fundamentally, $\mathrm{GPE}^*$ represents the (negative of) the vertical integral of "true" vertical normal stress $\bar\sigma_{zz}$. Meanwhile, the "true" $\mathrm{GPE}$ is the integral of the lithostatic pressure ($\bar P_L$). In general, $\mathrm{GPE}^* \neq \mathrm{GPE}$. The quantities are only equal when the vertical normal stress in both columns is lithostatic. This is not the case for trench topography!. The paper shows that one can interpret  $\bar\sigma_{zz}$ in as a GPE-like quantity, even for non-isostatic columns, if we use a corrected density ($\rho^*$) that accounts for horizontal gradients in the vertical shear stress: $\frac{\partial \tau_{zx}}{\partial x}$.


**Some further points:**

- It is the  only location of the columns in Equation **(1)** & **(2)** that differentiates the trench pull force, from e.g, ridge push; both are net horizontal force components that arise due to differences described by the quantity $\Delta \bar\sigma_{zz}$.

- In an idealised 2D subducting plate, these 2 contributions (ridge push and trench pull) combine in a non-overlapping way: the ridge push captures the net force due to the isostatic topography, while the trench pull captures the net force  due to the non-isostatic topography.

- the absolute value of $F_{D}$ has meaning: it defines (in an overall sense) whether the stress state is tending to extend $F_{D}>0$ or shorten $F_{D}<0$ the lithosphere. Because significant differential stress can be supported only where $z' < z'_m$, the value of $F_{D}$ (i.e., the integral) should converge to a constant value (approximately independent of $z_c$).

- the magnitude of $\mathrm{GPE}^*$ in a column has no direct significance in terms of the horizontal force balance - only differences between columns matter. Unlike $F_{D}$, magnitude of $\mathrm{GPE}^* = \bar\sigma_{zz}$ in a single column always increases if we choose a larger integration depth. However, the integral of the $\Delta \mathrm{GPE}^* = \Delta \bar\sigma_{zz}$ (difference between columns) should converge with integration depth - if the assumption about the negligible pressure (or vertical normal stress) gradients at the compensation level is valid.

- In the MDOODZ7.0 code, the convention is that $z$ is positive up. In this notebook, the $z$ coordinate is switched for consistency with the manuscript. In terms of the signs of the stress tensor components, the only other required change in sign is for the shear stress.

- the origin of the z coordinate in this analysis is the initial level of the rock surface interface, the upper limit of the vertical integrations is the top of the model domain ($z=$ -20 km), inclusive of the "air" layer. The region above the rock surface has zero density, so $(\rho_m - \rho_w) = \rho_m =$ 3250 $\text{kg m}^{-3}$.


**Proposed scaling relationship:**


In the manuscript, the following scaling relationship is proposed for the trench pull force:  

\begin{equation}
\text{trench pull force} \equiv \Delta \mathrm{GPE}^* = (\rho_m - \rho_w) g w_T ( \frac{z'_m}{2}) \approx \Delta P_T z'_{np}
\end{equation}

Where $w_T$ is the deflection at the trench axis relative to the isostatic level, $\Delta P_T$ is referred to as the trench pressure deficit, $z'_m$ is the mechanical thickness, and $z'_{np}$ is the neutral plane depth. 

This notebook evaluates this relationship based on the results of a numerical model. Because the numerical model provides access to the vertical normal stress $\sigma_{zz}$, we can calculate the trench pull force  directly.

# Part 2: Flexural support of trench topography and $\Delta \mathrm{GPE}^*$ estimate

In this section we:
* estimate the **flexurally supported component** of the topography,
* calculate the location of the **nearest isostatic column** to the trench, and
* estimate the **trench pull force** based on the scaling relationship defined in the manuscript.

In the classic thin plate flexure model, the vertical force balance is typically expressed as:

$\frac{dV}{dx} = -q$

where $V$ is the vertical shear stress resultant, and the normal load ($q$) is the given by the isostatic restoring force: 

\begin{equation}
q = (\rho_m - \rho_w) g w(x)
\end{equation}

Here, $V$ is approximated by integrating the **vertical component of the traction** $\tau_{zx}\cdot \hat n$ down to the assumed compensation level $z_c$, and the horizontal gradient ($\frac{dV}{dx}$) is approximated using the numpy gradient function. 


We can compare the actual **model topographic variation** in the plate seaward of the trench with the **topographic variation implied by the gradients in $V$**.

The condition $\frac{dV}{dx} = 0$ corresponds to a location where the column experiences no flexural support. Seaward of the trench, the first location where this condition is met is often called the first zero crossing, and is denoted $x_I$.



```python
#integrate shear stresses along z axis using trapezoidal rule 
#neg 1 accounts for the unit vector
Vx = -1*np.trapz(tau_xz_grid[vint_mask, 1:], dx=1e3, axis =0)
#gradient along z usign central difference
dVdx = np.gradient(Vx)
#calc the no-isostatic delection (w) supported by the estimate of dVdx
vtopo = dVdx/(delrho*g)
```

```python

#get the first zero crossing. Closest index appoach is sufficient here.
#use a reasonable window (meaning here, kms) to search for the crossing
window=500
indx = np.argmin(-1*np.sum(tau_xz_grid, axis=0)[vxpts > tloc][0:window])
xi_loc = vxpts[vxpts > tloc][0:window][indx]
xi_indx = np.argmin(np.abs(vxpts - xi_loc)) 
xi_dist = xi_loc - tloc
```

```python
#calculate the topographic difference between the first zero crossing and the trench,
#note that this is not the total topography between the trench and outer rise,
#in the paper this this is $w_T$

trench_relative_topo = topo[:,1][np.argmin(abs(topo[:,0] - xi_loc))] - topo[:,1][np.argmin(abs(topo[:,0] - tloc))]

#get the index of the mid point between trench and first zero crossing 
xm_indx = int(tindx +   0.5*(xi_indx- tindx))
```

```python
#this is a depth mask to limit the search for the neutral plane, 
#or the zero-crossing of the function tau_xx - tau_zz 
np_mask = np.logical_and(czpts>10e3, czpts<50e3)

# Extract the masked array and corresponding x-values
z_vals = diff_stress_grid[:, xm_indx][np_mask]
x_vals = czpts[np_mask]

# Find indices where the sign changes (zero crossing)
sign_changes = np.where(np.diff(np.sign(z_vals)))[0]

# If a zero crossing exists, interpolate
if len(sign_changes) > 0:
    idx = sign_changes[0]  # Take the first crossing
    x1, x2 = x_vals[idx], x_vals[idx + 1]
    y1, y2 = z_vals[idx], z_vals[idx + 1]

    # Linear interpolation for zero crossing
    np_depth = x1 - y1 * (x2 - x1) / (y2 - y1)
    print(f"Interpolated neutral plane crossing at z = {1e-3*np_depth:.1f} km")
else:
    print("No zero crossing found within the specified range.")
```

```python
#estimate of the `trench pull force' using the scaling relationship from the paper
gpe_est = trench_relative_topo*delrho*g*np_depth

print(f"Trench Relative Topo: {1e-3*trench_relative_topo:.2f} km")
print(f"Neutral plane depth: {1e-3*np_depth:.2f} km")
print(f"Trench pull force (scaling estimate): {gpe_est*1e-12:.2f} TN/m")
```

## Part 3: Horizontal force balance

We now use quadrature to estimate the vertically integrated terms that appear in the **vertically integrated horizontal force balance** (equations 1 - 3). All integrals are approximated using **trapezoidal integration**.






```python
#calculate the integrated qunatities using Trapezoidal function

# \Delta \bar \sig_{xx}$
#Note: this is the variation in the `full' horizontal normal stress (traction) resultant (see equation 1)
del_bar_sig_xx =  ( 1e-12*np.trapz(sig_xx_grid[cint_mask, :], x=czpts[cint_mask], axis=0) - 
                  1e-12*np.trapz(sig_xx_grid[cint_mask, :], x=czpts[cint_mask], axis=0)[tindx])


#\Delta \bar \F_{D}(x) = \Delta \bar{tau_(xx) - tau_(zz)}$
#The variation in the in-plane differential stress term
del_FD_x = (1e-12*np.trapz(diff_stress_grid[cint_mask, :], x=czpts[cint_mask], axis=0) -
           1e-12*np.trapz(diff_stress_grid[cint_mask, :], x=czpts[cint_mask], axis=0)[tindx])



#$\Delta {GPE^*} = - \bar \sig_{zz}$, 
#Note: that this quantity is NOT the true GPE diffence (i.e. not $\Delta  \bar P_L$),
del_GPE_x =-1*( 1e-12*np.trapz(sig_zz_grid[cint_mask, :], x=czpts[cint_mask], axis=0) - 
                1e-12*np.trapz(sig_zz_grid[cint_mask, :], x=czpts[cint_mask], axis=0)[tindx])


#basal shear term
# we intrgrate accross the entire horizontal domain and reference this to zero at the trench (x0)
FB_x = 1e-12*((cumtrapz(tau_xz_grid[vbase_indx,:], x=vxpts))[:] \
        -    (cumtrapz(tau_xz_grid[vbase_indx,:], x=vxpts))[:][tindx])

#upper shear term (should be zero do to free slip condition)
# we intrgrate accross the entire horizontal domain and reference this to zero at the trench (x0)
FT_x = -1e-12*((cumtrapz(tau_xz_grid[0,:], x=vxpts))[:] \
        -    (cumtrapz(tau_xz_grid[0,:], x=vxpts))[:][tindx])


# 2. Edge-averaged basal and upper shear (aligned with cell faces)
tau_basal_face = 0.5 * (tau_xz_grid[vbase_indx, :-1] + tau_xz_grid[vbase_indx, 1:])
tau_top_face   = 0.5 * (tau_xz_grid[0, :-1]          + tau_xz_grid[0, 1:])

FB_x = 1e-12 * (cumtrapz(tau_basal_face, x=cxpts) - cumtrapz(tau_basal_face, x=cxpts)[tindx])
FT_x = -1e-12 * (cumtrapz(tau_top_face,   x=cxpts) - cumtrapz(tau_top_face,   x=cxpts)[tindx])
FB_x = np.insert(FB_x, 0, 0.0)
FT_x = np.insert(FT_x, 0, 0.0)


#and this is the TRUE variation in the in-plane differential stress resultant
FD_x  = 1e-12*np.trapz(diff_stress_grid[cint_mask, :], x=czpts[cint_mask], axis=0) 
#This is the TRUE value of the in-plane differential stress resultant at the trench.
# we use this as a refernce point, as explained in the following section 
```

**Plot the _interesting_ form of the horizontal force balance**

We now want to assess the extended (interesting) form of the horizontal force balance: 

\begin{equation}
\Delta F_{D}  -  \Delta \mathrm{GPE}^*  +  F_B = 0 \;\; \mathrm{{\bf(a)}}
\end{equation}

In this form it is implied that all quantities will be set to zero at the reference location ($x_0$, e.g., the trench)

But we can also plot this in a slighly more insightful form:

Using our definition of the $\Delta$ symbol (see previous text cell) we can expand $\Delta F_{D}$ , and rewrite Equation 3a as:

\begin{equation}
F_{D}(x)  -  \left( \Delta \mathrm{GPE}(x)^* -  F_{D}(x_0)  \right) +  F_B(x) = 0 \;\;\mathrm{{\bf(3b)}}
\end{equation}


In other words, we can plot the actual $F_{D}(x)$, and use term ($-  F_{D}(x_0))$ to shift the value of $ \Delta \mathrm{GPE}(x)$. 

You might choose do this because $F_D(x)$ has a objective physical meaning at any given point x (i.e. for any lithpsheric column): It represents the tendency of the integrated in-plane differential stress to **extend** ($F_D> 0$) or **shorten** the lithosphere ($F_D<0$). 

If we plotted the variation in terms as described by (3) all terms would be zero at $x_0$ and we loose the information about the actual value of $F_{D}(x)$. 

Swtiching the `REF_VAL` parameter in the next cell will allow you to plot either form of the force balance

## Part 4 : Depth variation of stress

The following plots show the depth distribution of various quantities of interest at key points in the trench–outer slope region. 

The most important panel is **c** showing the difference in the vertical normal stress between $x_{I}$ and the trench ($x_T$). That is: 

\begin{equation}
\Delta \sigma_{zz}(z) = \sigma_{zz}(x_{I}, z) - \sigma_{zz}(x_{T}, z)
\end{equation}

When the manuscript talks about the "equilibration of vertical normal stress" it is referring to the fact that this difference diminishes with depth. The trench pull force is equal to the **area under** $\Delta \sigma_{zz}(z)$.

In the figure below, a smoothing operator is applied to reduce high-frequency noise arising from stress rotations in shear bands, which develop due to pressure-dependent plasticity. These small stress rotations introduce fluctuations in the vertical shear stress that propagate into the vertical normal stress, as the latter depends on horizontal gradients in vertical shear stress. The shear bands produce perturbations that oscillate around zero and have negligible effect on the vertically integrated values. The smoothing window, defined by `smoothwin`, averages adjacent grid cells (spaced 1 km apart) to suppress this noise. The magnitude of the trench pull force is primarily controlled by the depth over which $\sigma_{zz}(z)$ equilibrates, and while it is sensitive to the integrated horizontal gradients of vertical shear stress, the local perturbations caused by shear bands have little influence when integrated over depth.

```python
def pv_plane_from_h5(file, xleft=0, zbase =0):
    
    data   = file['/Model/Params']
    nx             = int(data[3])                   #/ No. grid points x
    nz             = int(data[4]) 
    
    xspan = data[1]
    zspan = data[2]
    
    center = (xspan/2. +  xleft, 
              zbase - zspan/2.,
              0)

    plane = pv.Plane(
        center=center,
        direction=(0, 0, -1),
        i_size=xspan,
        j_size=zspan,
        i_resolution=nx-1,
        j_resolution=nz-1,
    )
    
    return plane

def lines_from_points(points):
    """Given an array of points, make a line set"""
    poly = pv.PolyData()
    poly.points = points
    cells = np.full((len(points) - 1, 3), 2, dtype=np.int_)
    cells[:, 1] = np.arange(0, len(points) - 1, dtype=np.int_)
    cells[:, 2] = np.arange(1, len(points), dtype=np.int_)
    poly.lines = cells
    return poly
```
