# Mathematical background & analysis notes

In this analysis, we consider the horizontal force balance across a finite domain representing part of the trailing plate.

The domain is chosen such that its boundaries are parallel to the coordinate axes. In this configuration, the traction vector on each boundary has only one nonzero component aligned with the corresponding outward normal. For the vertical boundaries, the outward normal is horizontal, so the relevant traction component is $t_x = \sigma_{xx}$. For the basal boundary, where the normal is vertical, the traction component contributing to the horizontal force balance is $t_x = \tau_{xz}$.

As a result, the vector equation for static equilibrium reduces, upon integration over the 3 contrubuting domain boundaries, to a scalar equation for the x-component of the net force ($F_x$) per unit length:


\begin{equation}
\sum F_x = \bar \sigma_{xx}(x_1) - \bar \sigma_{xx}(x_0) + \int_{x_0}^{x_1} \tau_{xz}(x, z_c) dx = 0 \;\;\mathrm{{\bf(1a)}}
\end{equation}

The **overbars** in Equation **(1)** denote vertical integration between an upper equipotential surface—typically sea level, and a lower equipotential surface at the compensation depth ($z_c$). This depth is somewhat arbitrary, but it is commonly chosen such that horizontal gradients in pressure (or vertical normal stress) are negligible, i.e., under a hydrostatic approximation  $\bf{^1}$ . This is the standard assumption in thin-plate flexural models.


If we in $\Delta$ symbol is a shorthand used to represent the difference between the vertical integrals at points $x_0$ and $x_1$, we can simplify further: 


\begin{equation}
\Delta \bar \sigma_{xx}+ \int_{x_0}^{x_1} \tau_{xz}(x, z_c) dx = 0 \;\;\mathrm{{\bf(1b)}}
\end{equation}


Equation **(1)** can be interpreted as follows: the difference in integrated horizontal traction due to basal shear stress ($\tau_{xz}$) between $x_0$ and $x_1$, at $z = z_c$, balances the difference in the vertically integrated horizontal traction due to the normal stress $\sigma_{xx}$. 

$\bf{^1}$ Equation **(1)** remains valid even if pressure gradients exist along the basal boundary. However, a significant pressure gradient would implies dynamic support of the topography and complicate the interpretation of the terms. The analysis shows that dynamic topography in the subduction model is negligible - on teh horzontal sacke ofthe trench topotgrapht. This is consistent with the standard assumtpuin that trench topography anomaly is instead supported by vertical shear stresses within the lithosphere.


----------------

We can rewrite $\sigma_{xx}$ as: 

$\sigma_{xx} = \tau_{xx} + \sigma_I= \tau_{xx} + \sigma_{zz} - \tau_{zz}$

Where $\sigma_I = -P$ is the mean stress. Stress tensor quantities are assumed negative in compression. Substituting in Equation **(1)** gives:

\begin{equation}
\Delta (\overline{\tau_{xx}-\tau_{zz}})  +  \Delta \bar\sigma_{zz}  +  \int_{x_0}^{x_1} \tau_{xz}(x, z_c) = 0 \;\;\mathrm{{\bf(2)}}
\end{equation}

In Equation **(1)** & **(2)** a positive change in any of the quantities, from $x_0 \rightarrow x_1$ represents a net force to the right. For instance a positive change in $\Delta (\overline{\tau_{xx}-\tau_{zz}})$ represents a reduction in the integrated horizontal normal stress relative to the vertical, we can say that the in-plane stress is becoming less compressional in the positive $x$ direction.  

In the manuscript, these 3 terms are symbolised: 

\begin{equation}
\Delta F_{D}  -  \Delta \mathrm{GPE}^*  +  F_B = 0 \;\; \mathrm{{\bf(3)}}
\end{equation}

Where the definition imposed is: $\Delta \mathrm{GPE}^* = -\Delta \bar\sigma_{zz}$

The sign change implemented here makes the sign of $\Delta \mathrm{GPE}^* $ consistent with the definition for the "true" GPE (which is based on positive quantity - pressure, rather than a negative quantity, the vertical normal stress).  This means that a positive change in $\Delta \mathrm{GPE}^*$ from  $x_0 \rightarrow x_1$, now represents a net force to the left. The reason for the asterisk on the $\mathrm{GPE}^*$, is discussed in the paper. Fundamentally, $\mathrm{GPE}^*$ represents the (negative of) the vertical integral of "true" vertical normal stress $\bar\sigma_{zz}$. Meanwhile, the "true" $\mathrm{GPE}$ is the integral of the lithostatic pressure ($\bar P_L$). In general, $\mathrm{GPE}^* \neq \mathrm{GPE}$. The quantities are only equal when the vertical normal stress in both columns is lithostatic. This is not the case for trench topography!. The paper shows that one can interpret  $\bar\sigma_{zz}$ in as a GPE-like quantity, even for non-isostatic columns, if we use a corrected density ($\rho^*$) that accounts for horizontal gradients in the vertical shear stress: $\frac{\partial \tau_{zx}}{\partial x}$.


**Some further points:**

- It is the  only location of the columns in Equation **(1)** & **(2)** that differentiates the trench pull force, from e.g, ridge push; both are net horizontal force components that arise due to differences described by the quantity $\Delta \bar\sigma_{zz}$.

- In an idealised 2D subducting plate, these 2 contributions (ridge push and trench pull) combine in a non-overlapping way: the ridge push captures the net force due to the isostatic topography, while the trench pull captures the net force  due to the non-isostatic topography.

- the absolute value of $F_{D}$ has meaning: it defines (in an overall sense) whether the stress state is tending to extend $F_{D}>0$ or shorten $F_{D}<0$ the lithosphere. Because significant differential stress can be supported only where $z' < z'_m$, the value of $F_{D}$ (i.e., the integral) should converge to a constant value (approximately independent of $z_c$).

- the magnitude of $\mathrm{GPE}^*$ in a column has no direct significance in terms of the horizontal force balance - only differences between columns matter. Unlike $F_{D}$, magnitude of $\mathrm{GPE}^* = \bar\sigma_{zz}$ in a single column always increases if we choose a larger integration depth. However, the integral of the $\Delta \mathrm{GPE}^* = \Delta \bar\sigma_{zz}$ (difference between columns) should converge with integration depth - if the assumption about the negligible pressure (or vertical normal stress) gradients at the compensation level is valid.

- In the MDOODZ7.0 code, the convention is that $z$ is positive up. In this notebook, the $z$ coordinate is switched for consistency with the manuscript. In terms of the signs of the stress tensor components, the only other required change in sign is for the shear stress.

- the origin of the z coordinate in this analysis is the initial level of the rock surface interface, the upper limit of the vertical integrations is the top of the model domain ($z=$ -20 km), inclusive of the "air" layer. The region above the rock surface has zero density, so $(\rho_m - \rho_w) = \rho_m =$ 3250 $\text{kg m}^{-3}$.


**Proposed scaling relationship:**


In the manuscript, the following scaling relationship is proposed for the trench pull force:  

\begin{equation}
\text{trench pull force} \equiv \Delta \mathrm{GPE}^* = (\rho_m - \rho_w) g w_T ( \frac{z'_m}{2}) \approx \Delta P_T z'_{np}
\end{equation}

Where $w_T$ is the deflection at the trench axis relative to the isostatic level, $\Delta P_T$ is referred to as the trench pressure deficit, $z'_m$ is the mechanical thickness, and $z'_{np}$ is the neutral plane depth. 

This notebook evaluates this relationship based on the results of a numerical model. Because the numerical model provides access to the vertical normal stress $\sigma_{zz}$, we can calculate the trench pull force  directly.

```python


# ---------------------------------------------------------------------
# Mesh setup and geometric parameters
# ---------------------------------------------------------------------

# Nx, Nz = number of VERTICES in the x and z directions
# (each grid cell is bounded by 2 vertices ⇒ there are Nx−1 and Nz−1 cell centers)
Nx  = data[3].astype(int)
Nz  = data[4].astype(int)

# Nominal grid spacing (m)
# use ceil to avoid small floating-point rounding errors when reconstructing domain
dx  = np.ceil((xmax - xmin) / (Nx - 1))   # horizontal spacing (m)
dz  = np.ceil((zmax - zmin) / (Nz - 1))   # vertical spacing (m)

# Vertex coordinates (boundaries of cells)
vxpts = np.linspace(xmin, xmax, Nx)       # x-coordinates of vertices
vzpts = np.linspace(zmin, zmax, Nz)       # z-coordinates of vertices

# Cell-center coordinates (midpoints between adjacent vertices)
cxpts = 0.5 * (vxpts[1:] + vxpts[:-1])    # x-centers of cells  → length Nx−1
czpts = 0.5 * (vzpts[1:] + vzpts[:-1])    # z-centers of cells  → length Nz−1

# ---------------------------------------------------------------------
# Indices for reference depths / limits
# ---------------------------------------------------------------------

# Index of the vertical vertex just beyond the integration limit z_c_depth
# (used for selecting the deepest vertical integration slice)
#vbase_indx = np.argmin(np.abs(vzpts - z_c_depth))
vbase_indx = vertical_index(z_c_depth, vzpts)

# Index of the cell center nearest to z_c_depth
# (used when plotting or sampling quantities defined on cell centers)
cbase_indx = vertical_index(z_c_depth, czpts)
```

```python
# ---------------------------------------------------------------------
# Load and prepare topography array from HDF5 file
# ---------------------------------------------------------------------

# Attempt to read "x" and "z" coordinates for surface topography.
# Some model outputs store these under 'x_mark'/'z_mark' instead of 'x'/'z'.
try:
    topo_ = np.column_stack((
        h5file["/Topo"]["x_mark"][:],
        h5file["/Topo"]["z_mark"][:]
    ))
except KeyError:
    topo_ = np.column_stack((
        h5file["/Topo"]["x"][:],
        h5file["/Topo"]["z"][:]
    ))

# Initialise a 3-column array for convenience.
# Columns: [x, z, (reserved placeholder for later derived quantity)]
new_array = np.zeros((topo_.shape[0], 3))
new_array[:, :2] = topo_

# Sort rows by x-coordinate to ensure monotonic topography profile
topo = new_array[np.argsort(new_array[:, 0])]


# ---------------------------------------------------------------------
# Identify trench position and reference index
# ---------------------------------------------------------------------

# The trench is defined as the x-position of the topographic minimum
tloc = topo[:, 0][np.argmin(topo[:, 1])]

# Find the first mesh vertex located to the right of the trench
tindx = np.argmax(cxpts > tloc)


# ---------------------------------------------------------------------
# Interpolate topography to the model mesh
# ---------------------------------------------------------------------

# Interpolate the topographic surface to vertex x-points (vxpts)
# → topo_mesh has same length as cxpts, providing z-topo at each vertex
topo_mesh = np.interp(cxpts, topo[:, 0], topo[:, 1])
```

```python
# ---------------------------------------------------------------------
# Load and orient stress components from HDF5, assuming z is positive downward
# ---------------------------------------------------------------------

# Shear stress τ_xz (traction in x on face normal to z)
# Stored at vertices → flip vertically to ensure z increases downward
tau_xz_grid = -1 * np.flipud(h5file["/Vertices"]["sxz"][:].reshape(Nz, Nx))

# Pressure P (cell-centered) → flip vertically to match same orientation
pressure_grid = np.flipud(h5file["/Centers"]["P"][:].reshape(Nz-1, Nx-1))

# Mean (isotropic) stress: σ_I = −P
sigma_I_grid = -1 * pressure_grid                         # (Nz-1, Nx-1)

# Deviatoric normal stresses τ_xx and τ_zz = −τ_xx (plane strain assumption)
tau_xx_grid = np.flipud(h5file["/Centers"]["sxxd"][:].reshape(Nz-1, Nx-1))
tau_zz_grid = -1 * tau_xx_grid

# ---------------------------------------------------------------------
# Reconstruct total normal stresses at cell centers
# ---------------------------------------------------------------------

# Total σ_xx = σ_I + τ_xx
sig_xx_grid = sigma_I_grid + tau_xx_grid
# Total σ_zz = σ_I + τ_zz
sig_zz_grid = sigma_I_grid + tau_zz_grid

# For plane strain, τ_xx − τ_zz = 2 τ_xx
diff_stress_grid = tau_xx_grid - tau_zz_grid


# ---------------------------------------------------------------------
# load temperatrue grid
# ---------------------------------------------------------------------
temp_grid = np.flipud(h5file["/Centers"]['T'][:].reshape(Nz-1, Nx-1))
```

# Part 2: Flexural support of trench topography and $\Delta \mathrm{GPE}^*$ estimate

In this section we:
* estimate the **flexurally supported component** of the topography,
* calculate the location of the **nearest isostatic column** to the trench, and
* estimate the **trench pull force** based on the scaling relationship defined in the manuscript.

In the classic thin plate flexure model, the vertical force balance is typically expressed as:

$\frac{dV}{dx} = -q$

where $V$ is the vertical shear stress resultant, and the normal load ($q$) is the given by the isostatic restoring force: 

\begin{equation}
q = (\rho_m - \rho_w) g w(x)
\end{equation}

Here, $V$ is approximated by integrating the **vertical component of the traction** $\tau_{zx}\cdot \hat n$ down to the assumed compensation level $z_c$, and the horizontal gradient ($\frac{dV}{dx}$) is approximated using the numpy gradient function. 


We can compare the actual **model topographic variation** in the plate seaward of the trench with the **topographic variation implied by the gradients in $V$**.

The condition $\frac{dV}{dx} = 0$ corresponds to a location where the column experiences no flexural support. Seaward of the trench, the first location where this condition is met is often called the first zero crossing, and is denoted $x_I$.



```python
#cxpts.shape
```

```python
# ---------------------------------------------------------------------
# 1. Map vertex-based τ_zx (stored as tau_xz_grid) to vertical face centers
# ---------------------------------------------------------------------
tau_zx_face = vertices_to_vertical_faces(tau_xz_grid)  # (Nz-1, Nx)
#tau_zx_face_masked = tau_zx_face[cint_mask, :]         # apply vertical mask (above z_c_depth)

# ---------------------------------------------------------------------
# 2. Integrate τ_zx vertically to get cumulative horizontal traction (N/m)
#    Negative sign accounts for downward-pointing unit vector (z positive down)
# ---------------------------------------------------------------------
Vx = -1 * trapz(tau_zx_face, dx=dz, axis=0)  # integrate over z, result shape (Nx,)
Vx = 0.5 * (Vx [:-1] + Vx [1:])

# ---------------------------------------------------------------------
# 3. Compute horizontal gradient (∂Vx/∂x) using centered finite differences
#    This gives horizontal change in integrated shear traction (N/m²)
# ---------------------------------------------------------------------
dVdx = np.gradient(Vx, dx)

# ---------------------------------------------------------------------
# 4. Compute the "no-isostatic" vertical deflection supported by this stress gradient
#    v_topo = (1 / Δρg) * ∂Vx/∂x
# ---------------------------------------------------------------------
vtopo = dVdx / (delrho * g)

# ---------------------------------------------------------------------
# 5. Locate the first zero-crossing of the vertically integrated shear stress
# ---------------------------------------------------------------------
window = 500  # search range in km (or adjusted if vxpts in m)
mask_x = vxpts > tloc
tau_sum = np.sum(tau_zx_face, axis=0)[mask_x]
indx = np.argmin(-1 * tau_sum[:window])
xi_loc = vxpts[mask_x][:window][indx]
xi_indx = np.argmin(np.abs(vxpts - xi_loc))
xi_dist = xi_loc - tloc

# ---------------------------------------------------------------------
# 6. Compute trench-relative topographic difference (w_T in paper)
# ---------------------------------------------------------------------
trench_relative_topo = (
    topo[:, 1][np.argmin(np.abs(topo[:, 0] - xi_loc))] -
    topo[:, 1][np.argmin(np.abs(topo[:, 0] - tloc))]
)

# ---------------------------------------------------------------------
# 7. Find midpoint between trench and first zero-crossing (useful reference)
# ---------------------------------------------------------------------
xm_indx = int(tindx + 0.5 * (xi_indx - tindx))
```

```python
#this is a depth mask to limit the search for the neutral plane, 
#or the zero-crossing of the function tau_xx - tau_zz 
np_mask = np.logical_and(czpts>10e3, czpts<50e3)

# Extract the masked array and corresponding x-values
z_vals = diff_stress_grid[:, xm_indx][np_mask]
x_vals = czpts[np_mask]

# Find indices where the sign changes (zero crossing)
sign_changes = np.where(np.diff(np.sign(z_vals)))[0]

# If a zero crossing exists, interpolate
if len(sign_changes) > 0:
    idx = sign_changes[0]  # Take the first crossing
    x1, x2 = x_vals[idx], x_vals[idx + 1]
    y1, y2 = z_vals[idx], z_vals[idx + 1]

    # Linear interpolation for zero crossing
    np_depth = x1 - y1 * (x2 - x1) / (y2 - y1)
    print(f"Interpolated neutral plane crossing at z = {1e-3*np_depth:.1f} km")
else:
    print("No zero crossing found within the specified range.")
```

```python
#estimate of the `trench pull force' using the scaling relationship from the paper
gpe_est = trench_relative_topo*delrho*g*np_depth

print(f"Trench Relative Topo: {1e-3*trench_relative_topo:.2f} km")
print(f"Neutral plane depth: {1e-3*np_depth:.2f} km")
print(f"Trench pull force (scaling estimate): {gpe_est*1e-12:.2f} TN/m")
```

## Part 3: Horizontal force balance
We now use the stress tensor components from the numerical model to estimate the terms that appear in the **vertically integrated horizontal force balance** (equations 1 - 3).


First let's validate the basic form of the vertically-integrated force balance (i.e. Equation 1): 

\begin{equation}
\Delta \bar\sigma_{xx}(x) +  F_B(x) = 0 
\end{equation}

At the beginning of this notebook, this equation is derived for two arbitrary points ($x_0$, $x_1$), and simply states that these quantities must cancel — a condition of static equilibrium (horizontal force balance). The points are of course arbitrary, but our interest here is the effect of the trench (Earth's topographic minima) relative to an isostatic column in the trailing plate.

If we fix $x_0$ (e.g. at the trench) and allow $x_1$ to be a variable, then we can regard these 3 quantities in Equation 3 as **functions of** $x$.  In the plots that follow, we now define the $\Delta$ operator to represent $\Delta(f,x_0,x) = f(x) - f(x_0)$. 

...so that differences are always measured relative to the reference position x_0.

---

**Discrete formulation on the finite-difference grid**

In the discrete system, each column corresponds to a vertical stack of **cells** (indexed by $(k = 0, 1, \dots, N_{x-1}$)), bounded by **vertices** ($N_x$ in total).   Vertical integrations are performed on quantities defined at **faces**, so integrated (barred) quantities (like `bar_sig_xx`) are naturally stored at **vertices**.

To form the discrete difference, we define $\Delta(\bar{f})$ on the **cell centers** $ (N_{x-1})$ elements, such that the value of  
$\Delta(\bar{f})$ at cell $k$ corresponds to:

$$
\Delta(\bar{f})_k = \bar{f}_{k+1} - \bar{f}_{ref},
\quad \text{for } k = 0, 1, \dots, N_x-2
$$


In other words, each cell-centered value $\Delta(\bar{f})_k$  represents the difference in vertically integrated traction between the traction on vertex index $k+1$  and a reference location $k_{ref}$. The most obvious choice for $k_{ref}$ would be to choose one of the boundaires of the doman, i.e. $k_{ref}= (0, N_x)$. However, for analysing the variation on force components near the trench, it is more practical to us a reference point that is near zero at the trench. There are a few options that you can toggle with the `REF_TYPE` parameter in the next cell. 

---




```python
#-----------------------------------
#calculate the vertically integrated quantities
#-----------------------------------


# σ̄ₓₓ  — vertically integrated normal stress (Nz−1, Nx)
sig_xx_face = centers_to_vertical_faces(sig_xx_grid)
bar_sig_xx = np.cumsum(sig_xx_face * dz, axis=0)


# σ̄zz  ≡  −GPE*  — vertically integrated vertical normal stress
sig_zz_face = centers_to_vertical_faces(sig_zz_grid)
bar_sig_zz = np.cumsum(sig_zz_face * dz, axis=0)


# (τ̄ₓₓ − τ̄zz)  ≡  Fᴰ(x)  — vertically integrated differential stress

diff_stress_face = centers_to_vertical_faces(diff_stress_grid)
bar_diff_stress = np.cumsum(diff_stress_face * dz, axis=0)


#-----------------------------------
#calculate the basal shear term
#-----------------------------------

#   ∫ τₓz dx ≡ Fᴮ(x)   — horizontally integrated basal shear traction
tau_xz_face = vertices_to_horizontal_faces(tau_xz_grid)
FB_x_grid = np.cumsum(tau_xz_face  * dx, axis=1)


# ---------------------------------------------------------------------
# Reference level choice for horizontally integrated quantities
# ---------------------------------------------------------------------
# REF_TYPE = 0 → reference all quantities to zero at the trench
# REF_TYPE = 1 → reference σ̄zz (GPE*) to |τ̄xx − τ̄zz| at trench,
#                and set FB(x) = 0 at the trench
# ---------------------------------------------------------------------

REF_TYPE = 1  # valid options: 0 or 1
assert REF_TYPE in (0, 1), "REF_TYPE must be 0 (trench-zero) or 1 (GPE*-referenced)."

if REF_TYPE == 0:
    # Reference all vertically integrated Δ quantities to zero at the trench
    ref_index = tindx
    del_bar_sig_xx      = bar_sig_xx[:, 1:]      - bar_sig_xx[:, [ref_index]]
    del_bar_sig_zz      = bar_sig_zz[:, 1:]      - bar_sig_zz[:, [ref_index]]
    del_bar_diff_stress = bar_diff_stress[:, 1:] - bar_diff_stress[:, [ref_index]]
    FB_x_grid           = FB_x_grid[:]           - FB_x_grid[:, [ref_index]]

elif REF_TYPE == 1:
    # Reference Δσ̄zz (ΔGPE*) to |τ̄xx − τ̄zz| at trench
    # and plot |τ̄xx − τ̄zz| rather than  Δ|τ̄xx − τ̄zz|
    FB_x_grid           = FB_x_grid[:]           - FB_x_grid[:, [tindx]]
    del_bar_sig_zz      = (bar_sig_zz[:, 1:] - bar_sig_zz[:, [tindx]]
                           - bar_diff_stress[:, [tindx]])
    del_bar_diff_stress = bar_diff_stress[:, 1:]
    del_bar_sig_xx      = bar_sig_xx[:, 1:]      - bar_sig_xx[:, [tindx]]
    FB_x_grid           = FB_x_grid[:]           - FB_x_grid[:, [tindx]]

#for plotting
if REF_TYPE == 0:
    gpelabel = r'$\Delta \mathrm{GPE}^*(x)$' 
    FD_label = r'$\Delta F_{D}(x)$'
elif REF_TYPE == 1:
    gpelabel = r'$\Delta \mathrm{GPE}^*(x) - F_{D}(x_t)$' 
    FD_label = r'$F_{D}(x)$'
```

**Plot the _interesting_ form of the horizontal force balance**

We now want to look at the more _interesting_ form of the horizontal force balance, where the $\sigma_{xx}$ is decomposed in such a way as to highlight:

1. the pressure-component of the horizontal traction that is attrubutable to the depth distribution of the vertical normal tress ($\sigma_{zz}(z)$) (this is where topographic forces like swell push, ridge push and trench pull are expressed
2. the deviatoric-related component of the horizortal traction do to a finite differential stress resultant $(\tau_{xx} - \tau_{zz})$

In the symbolic form we write this as:

\begin{equation}
\Delta F_{D}  -  \Delta \mathrm{GPE}^*  +  F_B = 0 \;\; \mathrm{{\bf(3a)}}
\end{equation}

Note that for overall force balance on a section of lithopshere bounded by 2 columns, only the changes in $F_{D}$ and $\mathrm{GPE}^*$ are relavent - not teh absolute value. For any valid force balance, we can add an arbitary consnant to the $F_{D}$ and $\mathrm{GPE}^*$, and the equilibroum is sustained. 

We have defined the trench pull force as $\Delta \mathrm{GPE}^*$ between a column at the trench and isostatic column of the same age on the trailign plate. We are about to use stresses from the numerical model to evaluate this quantity. If we apply Equation 3a directly, it applies that changes in and of tehseqantitie are reference to (i.e zeroed) at the trench. 

However,  there is alternative way to plot the force balance, which is potentialy more insightfuk. 

Using our definition of the $\Delta$ symbol (see previous text cell) we can expand $\Delta F_{D}$ , and rewrite Equation 3a as:

\begin{equation}
F_{D}(x)  -  \left( \Delta \mathrm{GPE}(x)^* -  F_{D}(x_0)  \right) +  F_B(x) = 0 \;\;\mathrm{{\bf(3b)}}
\end{equation}


In other words, we can plot the actual $F_{D}(x)$, and use term ($-  F_{D}(x_0))$ to reference the value of $ \Delta \mathrm{GPE}(x)$. 

The advantage to doing it this was is that $F_D(x)$ has an objective physical meaning at any given lithospheric column: It represents the tendency of the integrated in-plane differential stress to **extend** ($F_D> 0$) or **shorten** the lithosphere ($F_D<0$). 

If we plotted the variation in terms as described by (3) all terms would be zero at $x_0$ and we loose the information about the actual value of $F_{D}(x)$. 

Switching the `REF_TYPE` parameter will allow you to plot either form of the force balance. Note that you you may need to change the vertical scale in the plot below.

## Part 4 : Depth variation of stress

The following plots show the depth distribution of various quantities of interest at key points in the trench–outer slope region. 

The most important panel is **c** showing the difference in the vertical normal stress between $x_{I}$ and the trench ($x_T$). That is: 

\begin{equation}
\Delta \sigma_{zz}(z) = \sigma_{zz}(x_{I}, z) - \sigma_{zz}(x_{T}, z)
\end{equation}

When the manuscript talks about the "equilibration of vertical normal stress" it is referring to the fact that this difference diminishes with depth. The trench pull force is equal to the **area under** $\Delta \sigma_{zz}(z)$.

In the figure below, a smoothing operator is applied to reduce high-frequency noise arising from stress rotations in shear bands, which develop due to pressure-dependent plasticity. These small stress rotations introduce fluctuations in the vertical shear stress that propagate into the vertical normal stress, as the latter depends on horizontal gradients in vertical shear stress. The shear bands produce perturbations that oscillate around zero and have negligible effect on the vertically integrated values. The smoothing window, defined by `smoothwin`, averages adjacent grid cells (spaced 1 km apart) to suppress this noise. The magnitude of the trench pull force is primarily controlled by the depth over which $\sigma_{zz}(z)$ equilibrates, and while it is sensitive to the integrated horizontal gradients of vertical shear stress, the local perturbations caused by shear bands have little influence when integrated over depth.


```python
# ---------------------------------------------------------------------
# Extract vertical profiles for key quantities along the trench–plate domain
# ---------------------------------------------------------------------

smoothwin = 10  # half-window (in grid columns) for local smoothing

# --- Vertical normal stress σzz ---
# At the isostatic column location (xi_indx)
# Average over ±smoothwin columns to reduce local numerical noise.
sig_zz_iso = (
    np.mean(sig_zz_grid[:, xi_indx - smoothwin : xi_indx + smoothwin], axis=1)
    if smoothwin else sig_zz_grid[:, xi_indx]
)

# At the trench column location (tindx)
sig_zz_trench = (
    np.mean(sig_zz_grid[:, tindx - smoothwin : tindx + smoothwin], axis=1)
    if smoothwin else sig_zz_grid[:, tindx]
)

# --- Other vertical profiles at the mid-point (xm_indx) ---
# These represent the mean thermal and mechanical structure halfway between
# the trench and the isostatic column.

# Temperature profile (°C)
temp_profile = (
    np.mean(temp_grid[:, xm_indx - smoothwin : xm_indx + smoothwin], axis=1)
    if smoothwin else temp_grid[:, xm_indx]
)

# In-plane differential stress τxx − τzz (Pa)
diff_stress_profile = (
    np.mean(diff_stress_grid[:, xm_indx - smoothwin : xm_indx + smoothwin], axis=1)
    if smoothwin else diff_stress_grid[:, xm_indx]
)

# Vertical shear stress τzx (Pa)
vertical_shear_stress_profile = (
    np.mean(tau_zx_face[:, xm_indx - smoothwin : xm_indx + smoothwin], axis=1)
    if smoothwin else tau_xz_face[:, xm_indx]
)
```

```python
def pv_plane_from_h5(file, xleft=0, zbase =0):
    
    data   = file['/Model/Params']
    nx             = int(data[3])                   #/ No. grid points x
    nz             = int(data[4]) 
    
    xspan = data[1]
    zspan = data[2]
    
    center = (xspan/2. +  xleft, 
              zbase - zspan/2.,
              0)

    plane = pv.Plane(
        center=center,
        direction=(0, 0, -1),
        i_size=xspan,
        j_size=zspan,
        i_resolution=nx-1,
        j_resolution=nz-1,
    )
    
    return plane

def lines_from_points(points):
    """Given an array of points, make a line set"""
    poly = pv.PolyData()
    poly.points = points
    cells = np.full((len(points) - 1, 3), 2, dtype=np.int_)
    cells[:, 1] = np.arange(0, len(points) - 1, dtype=np.int_)
    cells[:, 2] = np.arange(1, len(points), dtype=np.int_)
    poly.lines = cells
    return poly
```
